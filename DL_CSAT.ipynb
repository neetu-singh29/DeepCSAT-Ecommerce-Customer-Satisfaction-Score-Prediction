**📖 Project Summary:**
* In this project titled "**DeepCSAT – Ecommerce Customer Satisfaction Score Prediction**", the goal was to predict customer satisfaction scores (CSAT scores 0 to 5) based on historical customer support data in an e-commerce environment.

* The project followed a complete Machine Learning + Deep Learning pipeline, starting with Exploratory Data Analysis (EDA), Data Preprocessing, Feature Engineering, Visualization, Hypothesis Testing, and finally Deep Learning model building.

* The dataset included 85,907 customer support tickets and 20 features including categorical, numerical, and textual information (like customer remarks). Initial data exploration revealed missing values, outliers, high-cardinality ID columns, and imbalance in the target classes (more 4s and 5s).

* Missing values were handled using mode and mean imputations. Outliers in numeric fields were removed using IQR filtering. Textual fields were cleaned using NLP techniques such as contraction expansion, lowercasing, punctuation removal, stopwords removal, tokenization, and lemmatization. * Three hypothesis tests were performed:

 1. ANOVA showed significant CSAT differences across agent shifts.

 2. Product category differences in CSAT were significant.

 3. Pearson correlation showed a strong negative correlation between response delay and CSAT.

* After feature scaling and dimensionality reduction (PCA with 95% variance retained), we handled data imbalance using SMOTE technique. We then trained 3 different Deep Learning models: a basic ANN, an ANN with Dropout, and an ANN with BatchNormalization.

* The final model with BatchNormalization achieved around 88-89% accuracy on unseen test data, making it highly reliable for business decision making. The final model was saved in .h5 format and successfully reloaded for sanity testing.

* The DeepCSAT project delivers actionable insights for operational improvements and can easily be deployed on real-time customer feedback systems, enhancing customer satisfaction and business revenue.

### ✅ Hypothetical Statement –

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Answer Here:**

🎯 Topic: Impact of Agent Shift on CSAT Score

**Hypotheses:**

**Null Hypothesis (H₀)**: There is no significant difference in average CSAT scores across different agent shifts.

**Alternate Hypothesis (H₁):** There is a significant difference in CSAT scores between agent shifts.

#### 2. Perform an appropriate statistical test.
**Statistical Test: One-Way ANOVA**ANOVA (since we're comparing means across >2 groups)

````python
# Perform Statistical Test to obtain P-Value
from scipy.stats import f_oneway

# Grouping CSAT scores by Agent Shift
shift_groups = [group['CSAT Score'] for name, group in df.groupby('Agent Shift')]

# ANOVA test
f_stat, p_val = f_oneway(*shift_groups)
print(f"F-statistic: {f_stat}, ANOVA p-value: {p_val}")
````

##### Which statistical test have you done to obtain P-Value?

**Answer Here:**

✅ **One-Way ANOVA**

##### Why did you choose the specific statistical test?

**Answer Here:**

ANOVA is used when comparing means of more than two groups (multiple support channels) to see if any group differs significantly.

🧠 **Interpretation:**

* If p < 0.05: Reject H₀ → Agent shift affects satisfaction.

* If p ≥ 0.05: Fail to reject H₀ → No strong evidence that shifts matter.

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments

# Remove Outliers using IQR Method
for col in ['Item_price', 'connected_handling_time']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]
    #print(df.shape)
````

### 📍 DL Model - 4: ANN + Dropout, Batch Normalization, and L2 Regularization

````python
# --- Model Building ---
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.regularizers import l2

# Define the ANN model
model = Sequential()
model.add(Dense(256, input_dim=X_train.shape[1], activation='relu', kernel_regularizer=l2(0.01)))
model.add(BatchNormalization())
model.add(Dropout(0.3))
model.add(Dense(128, activation='relu', kernel_regularizer=l2(0.01)))
model.add(BatchNormalization())
model.add(Dropout(0.3))
model.add(Dense(64, activation='relu', kernel_regularizer=l2(0.01)))
model.add(BatchNormalization())
model.add(Dropout(0.2))
model.add(Dense(6, activation='softmax'))  # 6 classes for CSAT Score (0-5)

# Compile the model
model.compile(optimizer=Adam(learning_rate=0.001),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Early Stopping
early_stopping = EarlyStopping(patience=10, restore_best_weights=True)

# Train the model
history = model.fit(X_train, y_train, epochs=100, batch_size=32,
                    validation_split=0.1, callbacks=[early_stopping], verbose=1)

# Evaluate the model
loss, accuracy = model.evaluate(X_test, y_test, verbose=0)
print(f'Test Accuracy: {accuracy:.4f}')

# --- Model Evaluation ---
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Make predictions
y_pred = model.predict(X_test)
y_pred_classes = np.argmax(y_pred, axis=1)

# Print Classification Report
print("Classification Report:")
print(classification_report(y_test, y_pred_classes))

# Print Confusion Matrix
print("\nConfusion Matrix:")
cm = confusion_matrix(y_test, y_pred_classes)
print(cm)

# Visualize Confusion Matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix')
plt.ylabel('Actual Label')
plt.xlabel('Predicted Label')
plt.show()

# --- Hyperparameter Tuning ---
#from tensorflow.keras.wrappers.scikit_learn import KerasClassifier
from scikeras.wrappers import KerasClassifier
from sklearn.model_selection import GridSearchCV

# Function to create the model (for GridSearchCV)
def create_model(learning_rate=0.001, dropout_rate=0.2, units=128):
    model = Sequential()
    model.add(Dense(units, input_dim=X_train.shape[1], activation='relu'))
    model.add(Dropout(dropout_rate))
    model.add(Dense(units // 2, activation='relu'))
    model.add(Dropout(dropout_rate / 2))
    model.add(Dense(6, activation='softmax'))
    optimizer = Adam(learning_rate=learning_rate)
    model.compile(optimizer=optimizer, loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model
````
